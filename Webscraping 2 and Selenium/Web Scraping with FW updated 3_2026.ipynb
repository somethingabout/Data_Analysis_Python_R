```python


out = []
for page in range(1,5):

    url = f"https://laborskateshop.com/collections/decks?filter.v.availability=1&page={page}"
    driver.get(url)
    driver.implicitly_wait(10)
    buckets = driver.find_elements(By.XPATH, '/html/body/main/div[2]/div/div/div/div/ul/li')

    # Getting all data and URL's
    count = 1
    
    for bucket in buckets:
        text_data = bucket.text
        link = driver.find_element(By.XPATH, f'/html/body/main/div[2]/div/div/div/div/ul/li[{count}]/div/div/div[2]/div[1]/h3/a').get_attribute('href')
        
        data = {
            'text_data':text_data,
            'link':link
        }
        out.append(data)
        count = count+1
```

```python
df = pd.DataFrame(out)
df['price'] = df.text_data.str.split('\n').str[-1].str.split(' ').str[0].str.replace('$','').astype('float64')
df['name'] = df.text_data.str.split('\n').str[0]

df
```

# Getting URL's from a webpage 

```python
driver = webdriver.Chrome()
chrome_options.add_argument("--window-size=1900,1000") #pick a size and stick with it, it can change how code intreacts with the website.
driver = webdriver.Chrome(options = chrome_options)
#____________________________________________________________________________________________________#
yt_url = "https://www.youtube.com/@bonappetit/videos"
driver.get(yt_url)

driver.implicitly_wait(20)

vids_urls = []
video_urls = driver.find_elements(By.XPATH, './/*[@id="video-title-link"]')

for url in video_urls:
    href = url.get_attribute('href')
    vids_urls.append(href)

print(vids_urls)
print(' ')
print('ALl video links visable in the Json has been collected')
```
